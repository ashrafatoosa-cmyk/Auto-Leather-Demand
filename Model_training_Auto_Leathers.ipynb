```python
import pandas as pd
import numpy as np

# 1. Select core attributes relevant to demand forecasting & feature engineering
core_cols = [
    'id', 'created_at', 'car_make', 'model', 'type',
    'material_used', 'material_id', 'pattern', 'primary_colour',
    'total_amount', 'discount', 'status'
]
clean_df = df_jobs[core_cols].copy()

# 2. Date conversion and index sorting
clean_df['created_at'] = pd.to_datetime(clean_df['created_at'], errors='coerce')
# Drop rows where order timestamp is completely missing
clean_df = clean_df.dropna(subset=['created_at']).sort_values('created_at').reset_index(drop=True)

# 3. Clean categorical text: lowercase, strip whitespace
cat_cols = ['car_make', 'model', 'type', 'material_used', 'pattern', 'primary_colour', 'status']
for col in cat_cols:
    clean_df[col] = clean_df[col].astype(str).str.strip().str.title()
    clean_df[col] = clean_df[col].replace({'Nan': 'Unknown', 'None': 'Unknown', '': 'Unknown'})

# 4. Standardize numerical fields
clean_df['total_amount'] = pd.to_numeric(clean_df['total_amount'], errors='coerce').fillna(0.0)
clean_df['discount'] = pd.to_numeric(clean_df['discount'], errors='coerce').fillna(0.0)

# 5. Filter out canceled/invalid records if indicated in status (optional inspection)
print("Unique Statuses:", clean_df['status'].value_counts())
print("\nCleaned Shape:", clean_df.shape)
print("Date Range:", clean_df['created_at'].min(), "to", clean_df['created_at'].max())
print("\nMissing Values:")
print(clean_df.isnull().sum())

# Display sample
display(clean_df.head(5))
```

```python
import pandas as pd
import numpy as np

# 1. Standardize Car Brands
brand_map = {
    'Vw': 'Volkswagen',
    'Volkswagon': 'Volkswagen',
    'Merc': 'Mercedes',
    'Unknown': 'Other'
}
clean_df['car_brand_clean'] = clean_df['car_make'].replace(brand_map)

# Keep top brands, group rare ones into 'Other'
top_brands = clean_df['car_brand_clean'].value_counts().nlargest(6).index.tolist()
clean_df['car_brand_clean'] = clean_df['car_brand_clean'].apply(lambda x: x if x in top_brands else 'Other')

# 2. Date Setup & Weekly Aggregation
clean_df['order_date'] = clean_df['created_at'].dt.date
clean_df['order_date'] = pd.to_datetime(clean_df['order_date'])

# Weekly time-series aggregation
weekly_df = clean_df.resample('W-MON', on='order_date').agg(
    total_sales=('id', 'count'),               # Volume: total seat covers made
    avg_discount=('discount', 'mean'),         # Promotional impact
    total_revenue=('total_amount', 'sum')      # Revenue tracking
).reset_index()

# Fill missing weeks with 0 sales
weekly_df['total_sales'] = weekly_df['total_sales'].fillna(0)
weekly_df['avg_discount'] = weekly_df['avg_discount'].fillna(0)
weekly_df['total_revenue'] = weekly_df['total_revenue'].fillna(0)

# 3. Calendar & Event Feature Engineering
# Christmas & New Year Window (Dec 15 to Jan 5)
weekly_df['is_holiday_season'] = weekly_df['order_date'].apply(
    lambda d: 1 if (d.month == 12 and d.day >= 15) or (d.month == 1 and d.day <= 7) else 0
)

# UK COVID-19 Lockdown Shock (March 2020 - July 2021)
weekly_df['covid_lockdown'] = weekly_df['order_date'].apply(
    lambda d: 1 if pd.Timestamp('2020-03-23') <= d <= pd.Timestamp('2021-07-19') else 0
)

# 4. Temporal 80/20 Split (Strict chronological order, no random shuffle)
split_idx = int(len(weekly_df) * 0.80)
train_df = weekly_df.iloc[:split_idx].copy()
test_df = weekly_df.iloc[split_idx:].copy()

print(f"Total Weeks: {len(weekly_df)}")
print(f"Train Weeks (80%): {len(train_df)} | From {train_df['order_date'].min().date()} to {train_df['order_date'].max().date()}")
print(f"Test Weeks (20%): {len(test_df)} | From {test_df['order_date'].min().date()} to {test_df['order_date'].max().date()}")

# Preview aggregated data
display(weekly_df.head(6))
```

```python
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
import matplotlib.pyplot as plt

# 1. Check Stationarity with Augmented Dickey-Fuller (ADF) Test
adf_result = adfuller(train_df['total_sales'])
print("--- Augmented Dickey-Fuller Test ---")
print(f"ADF Statistic: {adf_result[0]:.4f}")
print(f"p-value: {adf_result[1]:.4f}")
if adf_result[1] <= 0.05:
    print("Series is stationary (p <= 0.05). Differencing (d=0) may be sufficient.")
else:
    print("Series is non-stationary (p > 0.05). Differencing (d=1) is recommended.")

# 2. Define Endogenous (target) and Exogenous (features)
exog_cols = ['is_holiday_season', 'covid_lockdown', 'avg_discount']

y_train = train_df['total_sales']
X_train = train_df[exog_cols]

y_test = test_df['total_sales']
X_test = test_df[exog_cols]

# 3. Fit SARIMAX Model with order=(1, 1, 1) and seasonal_order=(1, 0, 1, 52)
# Using weekly seasonality (52 weeks/year)
sarimax_model = sm.tsa.SARIMAX(
    y_train,
    exog=X_train,
    order=(1, 1, 1),
    seasonal_order=(1, 0, 1, 52),
    enforce_stationarity=False,
    enforce_invertibility=False
)

sarimax_fit = sarimax_model.fit(disp=False)
print("\n--- SARIMAX Model Summary ---")
print(sarimax_fit.summary())

# 4. Predict on Test Set (Out-of-sample 75 weeks)
sarimax_forecast = sarimax_fit.get_forecast(steps=len(test_df), exog=X_test)
test_df['sarimax_pred'] = sarimax_forecast.predicted_mean.values
# Ensure no negative predictions (sales volume cannot be below zero)
test_df['sarimax_pred'] = test_df['sarimax_pred'].clip(lower=0)

# 5. Quick visual check of test vs predicted
plt.figure(figsize=(12, 5))
plt.plot(test_df['order_date'], test_df['total_sales'], label='Actual Sales', color='black')
plt.plot(test_df['order_date'], test_df['sarimax_pred'], label='SARIMAX Prediction', color='blue', linestyle='--')
plt.title('Out-of-Sample Test Set: Actual vs SARIMAX Forecast')
plt.xlabel('Date')
plt.ylabel('Weekly Seat Covers Sold')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

# 1. Feature selection and scaling
features = ['total_sales', 'is_holiday_season', 'covid_lockdown', 'avg_discount']

scaler = MinMaxScaler(feature_range=(0, 1))
scaled_train = scaler.fit_transform(train_df[features])
scaled_test = scaler.transform(test_df[features])

# 2. Convert time series into sliding window sequences
def create_sequences(data, window_size=8):
    X, y = [], []
    for i in range(len(data) - window_size):
        X.append(data[i:i + window_size, :])
        y.append(data[i + window_size, 0])  # Predict total_sales (index 0)
    return np.array(X), np.array(y)

WINDOW_SIZE = 8

# Combine end of train with test to keep the continuous test set length intact
full_scaled = np.vstack([scaled_train, scaled_test])
X_all, y_all = create_sequences(full_scaled, window_size=WINDOW_SIZE)

# Separate back into train and test sequences
X_train_seq = X_all[:len(train_df) - WINDOW_SIZE]
y_train_seq = y_all[:len(train_df) - WINDOW_SIZE]

X_test_seq = X_all[len(train_df) - WINDOW_SIZE:]
y_test_seq = y_all[len(train_df) - WINDOW_SIZE:]

# 3. Build LSTM Architecture
tf.random.set_seed(42)
lstm_model = Sequential([
    LSTM(64, activation='relu', return_sequences=True, input_shape=(WINDOW_SIZE, len(features))),
    Dropout(0.2),
    LSTM(32, activation='relu'),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(1)
])

lstm_model.compile(optimizer='adam', loss='mse')
lstm_model.summary()

# 4. Train the model
history = lstm_model.fit(
    X_train_seq, y_train_seq,
    epochs=50,
    batch_size=16,
    validation_split=0.1,
    verbose=1
)

# 5. Inverse Transform Predictions
lstm_pred_scaled = lstm_model.predict(X_test_seq)

# Build dummy array to invert scaling specifically for total_sales
dummy_array = np.zeros((len(lstm_pred_scaled), len(features)))
dummy_array[:, 0] = lstm_pred_scaled.flatten()
inv_pred = scaler.inverse_transform(dummy_array)[:, 0]

# Attach predictions to test_df
test_df['lstm_pred'] = inv_pred
test_df['lstm_pred'] = test_df['lstm_pred'].clip(lower=0)

# 6. Plot Test Set: Actual vs SARIMAX vs LSTM
plt.figure(figsize=(12, 5))
plt.plot(test_df['order_date'], test_df['total_sales'], label='Actual Sales', color='black')
plt.plot(test_df['order_date'], test_df['sarimax_pred'], label='SARIMAX Prediction', color='blue', linestyle='--')
plt.plot(test_df['order_date'], test_df['lstm_pred'], label='LSTM Prediction', color='red')
plt.title('Out-of-Sample Test Set: Actual vs SARIMAX vs LSTM')
plt.xlabel('Date')
plt.ylabel('Weekly Seat Covers Sold')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd

# 1. Error Metrics Calculation
def calculate_wape(actual, predicted):
    return (np.sum(np.abs(actual - predicted)) / np.sum(actual)) * 100

def calculate_mape(actual, predicted):
    # Avoid zero division by filtering out zero sales points
    mask = actual > 0
    return np.mean(np.abs((actual[mask] - predicted[mask]) / actual[mask])) * 100

# Evaluation on 75 unseen test weeks
actual = test_df['total_sales'].values
sarimax_pred = test_df['sarimax_pred'].values
lstm_pred = test_df['lstm_pred'].values

metrics_summary = pd.DataFrame({
    'Metric': ['WAPE (%)', 'MAPE (%)'],
    'SARIMAX': [calculate_wape(actual, sarimax_pred), calculate_mape(actual, sarimax_pred)],
    'LSTM': [calculate_wape(actual, lstm_pred), calculate_mape(actual, lstm_pred)]
})

print("--- MODEL PERFORMANCE BENCHMARK ---")
display(metrics_summary)

# 2. Business Analytics: Brand & Color breakdown
print("\n--- TOP 5 CAR BRANDS (TOTAL SHARE) ---")
brand_counts = clean_df['car_brand_clean'].value_counts()
brand_share = (brand_counts / brand_counts.sum()) * 100
display(pd.DataFrame({'Orders': brand_counts, 'Share (%)': brand_share.round(2)}).head(6))

print("\n--- TOP 5 SEAT COVER COLOURS ---")
color_counts = clean_df['primary_colour'].value_counts()
color_share = (color_counts / color_counts.sum()) * 100
display(pd.DataFrame({'Orders': color_counts, 'Share (%)': color_share.round(2)}).head(5))

# 3. Holiday and Promotional Elasticity
holiday_avg = weekly_df.groupby('is_holiday_season')['total_sales'].mean()
print("\n--- AVERAGE WEEKLY ORDERS: HOLIDAY VS REGULAR ---")
print(f"Regular Weeks: {holiday_avg.get(0, 0):.1f} units")
print(f"Holiday Weeks: {holiday_avg.get(1, 0):.1f} units")

clean_df['has_discount'] = clean_df['discount'] > 0
discount_split = clean_df['has_discount'].value_counts(normalize=True) * 100
print(f"\nOrders placed with promotional discount: {discount_split.get(True, 0):.2f}%")
```

```python
import pandas as pd
import numpy as np

# 1. Export actual vs predicted for dashboard visual plots
export_forecast = test_df[['order_date', 'total_sales', 'sarimax_pred', 'lstm_pred']].copy()
export_forecast.columns = ['Date', 'Actual_Demand', 'SARIMAX_Forecast', 'LSTM_Forecast']
export_forecast.to_csv('dashboard_forecast_comparison.csv', index=False)

# 2. Export Brand breakdown
brand_summary = pd.DataFrame({
    'Brand': clean_df['car_brand_clean'].value_counts().index,
    'Total_Orders': clean_df['car_brand_clean'].value_counts().values,
    'Market_Share_Pct': (clean_df['car_brand_clean'].value_counts(normalize=True).values * 100).round(2)
})
brand_summary.to_csv('dashboard_brand_insights.csv', index=False)

# 3. Export Performance Metric Summary Table
metrics_summary.to_csv('dashboard_model_metrics.csv', index=False)

# 4. Generate next 4-week future forecast with trained LSTM
last_sequence = full_scaled[-WINDOW_SIZE:].reshape(1, WINDOW_SIZE, len(features))
future_predictions = []

curr_seq = last_sequence.copy()
for step in range(4):
    next_pred_scaled = lstm_model.predict(curr_seq, verbose=0)[0, 0]

    # Invert scale
    dummy = np.zeros((1, len(features)))
    dummy[0, 0] = next_pred_scaled
    unscaled_val = scaler.inverse_transform(dummy)[0, 0]
    future_predictions.append(max(0, round(unscaled_val)))

    # Construct next step input vector (assume non-holiday, no lockdown, avg discount)
    next_step_input = np.array([next_pred_scaled, 0.0, 0.0, train_df['avg_discount'].mean()])
    curr_seq = np.append(curr_seq[:, 1:, :], next_step_input.reshape(1, 1, len(features)), axis=1)

# Future dates setup
last_date = test_df['order_date'].max()
future_dates = [last_date + pd.Timedelta(weeks=i+1) for i in range(4)]
future_df = pd.DataFrame({
    'Forecast_Week': [d.strftime('%Y-%m-%d') for d in future_dates],
    'Projected_Seat_Covers': future_predictions
})

print("--- NEXT 4 WEEKS OPERATIONAL FORECAST ---")
display(future_df)
future_df.to_csv('dashboard_future_month_forecast.csv', index=False)
```
